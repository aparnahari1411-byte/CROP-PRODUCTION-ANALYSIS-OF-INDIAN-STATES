<h3 style="color:#A93226; margin-top:20px;">
9) Handling Outliers
</h3>

<p>
Outliers are extreme values that differ significantly from other data points. 
They can affect the accuracy of analysis and visualizations, so identifying and handling them is important.
</p>

<p>
<b>Explanation:</b> A boxplot was used to visualize the distribution of numerical data and identify the presence of outliers. 
It provides a clear summary of the data by showing the median, spread, and variability, along with extreme values that lie outside the normal range. 

By using a boxplot, it becomes easy to detect unusually high or low values that may affect the accuracy of analysis. 
This visual representation helps in deciding whether outliers need to be removed or treated before performing further analysis.
</p>

```python

import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=df['production'])
plt.title("Boxplot of Production")
plt.show()
```

<p>
<b>Insight:</b> The boxplot shows that most production values are concentrated at lower levels, while a significant number of extreme values are present on the higher side. 
This indicates a right-skewed distribution, where a few records have very high production compared to the majority. 
These extreme values are considered outliers and can influence the overall analysis.
</p>

<h4 style="color:#A93226;  margin-top:15px;">
Removing Outliers Using Percentile Method
</h4>

<p>
Extreme values were removed by filtering the top 1% of production values using percentile-based filtering.
</p>

```python
upper_limit = df['production'].quantile(0.99)

df = df[df['production'] <= upper_limit]
```

```python
df['production'].describe()
```

```python
sns.boxplot(x=df['production'])
plt.title("Boxplot After Removing Outliers")
plt.show()
```

<p>
<b>Insight:</b> After removing outliers, the distribution of production values became more controlled and less influenced by extreme values. 
The boxplot shows that most values are concentrated within a smaller range, indicating improved data consistency. 
However, the data still exhibits slight right skewness, which suggests that some regions or crops naturally have higher production levels. 
This cleaned dataset is more suitable for accurate analysis and visualization.
</p>

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15,4))

plt.subplot(1,3,1)
sns.boxplot(x=df['Area'], color='skyblue')
plt.title("Area Before Outlier Removal")

plt.subplot(1,3,3)
sns.boxplot(x=df['Yield'], color='purple')
plt.title("Yield Before Outlier Removal")

plt.show()
```

```python
df_clean = df.copy()

df_clean = df_clean[
    (df_clean['Area'].between(*[
        df['Area'].quantile(0.25) - 1.5*(df['Area'].quantile(0.75)-df['Area'].quantile(0.25)),
        df['Area'].quantile(0.75) + 1.5*(df['Area'].quantile(0.75)-df['Area'].quantile(0.25))
    ])) &
    
    (df_clean['Yield'].between(*[
        df['Yield'].quantile(0.25) - 1.5*(df['Yield'].quantile(0.75)-df['Yield'].quantile(0.25)),
        df['Yield'].quantile(0.75) + 1.5*(df['Yield'].quantile(0.75)-df['Yield'].quantile(0.25))
    ]))
]
```

```python
plt.figure(figsize=(15,4))

plt.subplot(1,3,1)
sns.boxplot(x=df_clean['Area'], color='green')
plt.title("Area After Outlier Removal")

plt.subplot(1,3,3)
sns.boxplot(x=df_clean['Yield'], color='blue')
plt.title("Yield After Outlier Removal")

plt.show()
```

<h3 style="color:#00008B;">
3. Production Trend Across States Over Years (Multivariate Line Plot)
</h3>

This graph is used to analyze how production changes over time across different states. A line plot helps in identifying trends, growth patterns, and comparisons between multiple states simultaneously.

```python

# Select top 5 states to avoid clutter
top_states = df_clean.groupby('State_Name')['Production'].sum().sort_values(ascending=False).head(5).index

# Filter dataset
df_top = df_clean[df_clean['State_Name'].isin(top_states)]

# Group by Year and State
trend_data = df_top.groupby(['Year', 'State_Name'])['Production'].sum().unstack()

# Plot
plt.figure(figsize=(14,6))

for state in trend_data.columns:
    plt.plot(trend_data.index, trend_data[state], marker='o', label=state)

plt.title("Production Trend Across Top States Over Years", fontsize=14)
plt.xlabel("Year")
plt.ylabel("Production")

plt.xticks(trend_data.index[::2], rotation=45)  # 🔥 key fix

plt.legend(title="State")
plt.grid(True)

plt.tight_layout()
plt.show()
```
